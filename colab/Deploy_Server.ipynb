# 🌙 Al-Ansar، آن لائن سرور ایک بار انسٹال کریں

یہ کام **صرف ایک بار** ہوتا ہے۔ اس کے بعد ایپ ہر بار خود سرور سے جڑ جاتی ہے، Colab کی ضرورت نہیں رہتی۔

**پہلے (ایک بار):**
1. [modal.com](https://modal.com) پر مفت اکاؤنٹ بنائیں (ہر ماہ $30 کا مفت کریڈٹ ملتا ہے، پورا کریڈٹ لینے کے لیے ان کی سیٹنگ میں کارڈ شامل کرنا پڑ سکتا ہے)۔
2. [ٹوکن کا صفحہ](https://modal.com/settings/tokens) کھول کر نیا ٹوکن بنائیں، اور اس کا **Token ID** اور **Token Secret** کاپی کریں۔
3. نیچے والے سیل میں دونوں چسپاں کریں، پھر **▶** دبائیں اور 10 سے 20 منٹ انتظار کریں (صرف پہلی بار)۔

> نوٹ بک میں ٹوکن لکھنے کے بعد اسے کسی کو شیئر نہ کریں۔ آپ کو GPU یہاں نہیں چاہیے، عام Colab چلے گا۔

```python
#@title ▶ سرور انسٹال کریں (Modal)
GITHUB_USER = "salmanabbasinm-hash"  #@param {type:"string"}
REPO = "AlAnsar-Voice-Cloning"        #@param {type:"string"}
MODAL_TOKEN_ID = ""      #@param {type:"string"}
MODAL_TOKEN_SECRET = ""  #@param {type:"string"}

import os, re, sys, subprocess
def sh(c): subprocess.run(c, shell=True, check=True)

if not MODAL_TOKEN_ID.strip() or not MODAL_TOKEN_SECRET.strip():
    raise RuntimeError("❌ پہلے Modal کا Token ID اور Token Secret اوپر ڈالیں")

print("⏳ تیاری...")
sh("pip -q install 'modal>=1.0'")
sh("rm -rf /content/app && git clone --depth 1 https://github.com/%s/%s /content/app" % (GITHUB_USER, REPO))
env = dict(os.environ, MODAL_TOKEN_ID=MODAL_TOKEN_ID.strip(), MODAL_TOKEN_SECRET=MODAL_TOKEN_SECRET.strip())
print("⏳ سرور بن رہا ہے (پہلی بار 10 سے 20 منٹ، اس سیل کو بند نہ کریں)...")
p = subprocess.run([sys.executable, "-m", "modal", "deploy", "/content/app/backend/modal_app.py"],
                   env=env, capture_output=True, text=True)
out = (p.stdout or "") + (p.stderr or "")
print(out[-2500:])
urls = re.findall(r"https://[A-Za-z0-9.-]+\.modal\.run", out)
if p.returncode != 0 or not urls:
    raise RuntimeError("❌ سرور نہیں بن سکا۔ اوپر کا آخری متن (لاگ) تصویر کی صورت میں بھیج دیں")
url = urls[-1]
from IPython.display import display, HTML
display(HTML('<div style="font-size:18px;line-height:2;direction:rtl">✅ <b>سرور تیار ہے</b><br>'
             'اپنا سرور کا لنک کاپی کر کے مجھے بھیج دیں، یا ایپ کی سیٹنگز ← سرور میں ڈالیں:<br>'
             '<input style="width:100%%;direction:ltr" value="%s" onclick="this.select()"></div>' % url))
```

### ضروری باتیں
- **لاگت:** Modal صرف استعمال کے سیکنڈوں کے پیسے لیتا ہے۔ مفت کریڈٹ سے عموماً مہینے میں سینکڑوں آڈیو بن سکتی ہیں، لیکن اس کی گنتی Modal کے ڈیش بورڈ میں دیکھتے رہیں اور وہاں خرچ کی حد (spending limit) لگا دیں۔
- **سویا ہوا سرور:** کچھ دیر استعمال نہ ہو تو سرور سو جاتا ہے، اور اگلی بار جاگنے میں 1 سے 2 منٹ لگتے ہیں۔
- **لنک:** جس کے پاس سرور کا لنک ہو وہ آپ کا کریڈٹ استعمال کر سکتا ہے۔ اس لیے اسے عام نہ کریں۔
